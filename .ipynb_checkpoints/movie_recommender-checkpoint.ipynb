```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python

def preprocess(text):
    text = text.lower()
    text = text.translate(str.maketrans('', '', string.punctuation))
    
    words = text.split()
    words = [w for w in words if w not in stop_words]
    
    words = [lemmatizer.lemmatize(w) for w in words]
    
    return " ".join(words)
```

```python
movies['movie title - year'] = movies['movie title - year'].apply(preprocess)
movies['expanded-genres'] = movies['expanded-genres'].apply(preprocess)
movies['clean-description'] = movies['description'].apply(preprocess)
```

```python
movies.head()
```

```python
movies['expanded-genres']= movies['expanded-genres'].apply(lambda x:x.split())
movies['clean-description']= movies['clean-description'].apply(lambda x:x.split())
```

```python
movies.head()
```

```python
#remove space
#flaming ear 1992 --> flamingear1992
movies['movie title - year'] = movies['movie title - year'].apply(lambda x: x.replace(" ", ""))
movies.head()        
```

```python
movies['tags']= movies['clean-description'] + movies['expanded-genres']
```

```python
new_df= movies[['movie title - year','rating', 'tags']]
```

```python
new_df.head()
```

```python
#convert list to string
new_df['tags'] = new_df['tags'].apply(lambda x: " ".join(x))
```

```python
new_df.head()
```

```python
new_df = new_df.sample(5000)
```

```python
#Vectorization 
from sklearn.feature_extraction.text import TfidfVectorizer

v = TfidfVectorizer(max_features=3000) 
transformed_output = v.fit_transform(new_df['tags']).toarray()
```

```python
transformed_output
```

```python
transformed_output.shape
```

```python
for i in new_df['movie title - year'].head(10):
    print(repr(i))
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# similarity = cosine_similarity(transformed_output)
```

```python

def recommend(movie):
    
    movie = movie.lower()
    
    # Find partial match
    matches = new_df[new_df['movie title - year'].str.contains(movie)]
    
    if matches.empty:
        print("Movie not found")
        return
    
    index = matches.index[0]
    
    # Use on-demand similarity (better for memory)
   sim_scores = cosine_similarity(transformed_output[index], transformed_output)
    distances = sorted(
        list(enumerate(sim_scores[0])),
        key=lambda x: x[1],
        reverse=True
    )
    
    for i in distances[1:6]:
        print(new_df.iloc[i[0]]['movie title - year'])
```

```python
recommend('flat2010')
```
